```python
%%writefile app.py
import os
import streamlit as st
import fitz  # PyMuPDF for PDF text extraction
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.embeddings.openai import OpenAIEmbeddings
from langchain.vectorstores import FAISS
from langchain.chains import RetrievalQA
from langchain.chat_models import ChatOpenAI  # Corrected model usage

# Set OpenAI API Key
OPENAI_API_KEY = "Open_ai_key"
os.environ["OPENAI_API_KEY"] = OPENAI_API_KEY

# Streamlit UI
st.title("Simple RAG System - PDF Q&A")

# Upload PDF
uploaded_file = st.file_uploader("Upload a PDF file", type="pdf")

if uploaded_file:
    st.write("Processing file...")

    # Read PDF content
    def extract_text_from_pdf(uploaded_file):
        doc = fitz.open(stream=uploaded_file.read(), filetype="pdf")
        text = ""
        for page in doc:
            text += page.get_text("text") + "\n"
        return text

    document_text = extract_text_from_pdf(uploaded_file)

    # Split text into chunks
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
    text_chunks = text_splitter.split_text(document_text)

    # Generate embeddings and store in FAISS
    embeddings = OpenAIEmbeddings()
    vectorstore = FAISS.from_texts(texts=text_chunks, embedding=embeddings)

    # Create Retrieval Chain
    llm = ChatOpenAI(temperature=0)  # Fix: Use ChatOpenAI
    retriever = vectorstore.as_retriever(search_type="similarity", search_kwargs={"k": 3})

    # Fix: Properly initialize RetrievalQA
    qa_chain = RetrievalQA.from_chain_type(llm=llm, retriever=retriever, chain_type="stuff")

    st.success("File processed! You can now ask questions.")

    # User input for Q&A
    query = st.text_input("Ask a question about the PDF:")

    if query:
        response = qa_chain.run(query)
        st.write("**Answer:**", response)
```
